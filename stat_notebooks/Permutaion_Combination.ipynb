# The Factorial

## 1. Core Intuition
A factorial calculates the total number of unique ways to arrange an entire group of distinct items into a linear sequence. 

##  2. Mathematical Definition
For any positive whole number $n$, the factorial (denoted by $n!$) is the product of all descending positive integers from $n$ down to $1$:
$$n! = n \times (n-1) \times (n-2) \times \dots \times 3 \times 2 \times 1$$

## 3. Edge Cases & Special Rules
* **The Zero Rule ($0! = 1$):** By algebraic pattern definition ($\frac{1!}{1} = 1$) and combinatorial logic (there is exactly $1$ way to arrange a set of zero objects: doing nothing), $0!$ evaluates to $1$.
* **Negative Integers ($(-n)! = \text{Undefined}$):** Extending the division-by-zero sequence to find negative integers yields $(-1)! = \frac{0!}{0} = \frac{1}{0}$. Because division by zero is mathematically impossible, factorials for negative integers are strictly undefined.


##  Glimpse of the Gamma Function

### 1. Core Intuition
The traditional factorial is discrete, meaning it only works for individual dots on a graph (whole numbers). The Gamma Function acts as a continuous "connect-the-dots" curve that bridges the gaps between whole numbers, allowing mathematicians to compute factorials for fractions, decimals, and complex numbers.

### 2. Mathematical Definition
For values where the real part of $n$ is positive, the Gamma function ($\Gamma$) is defined via a calculus integral that scales to infinity:
$$\Gamma(n) = \int_{0}^{\infty} x^{n-1} e^{-x} \, dx$$

Because of how the calculus simplifies, the function is shifted from the standard factorial by exactly $1$:
$$n! = \Gamma(n+1) \quad \text{or} \quad \Gamma(n) = (n-1)!$$

### 3. Edge Cases & Special Rules
* **Fractional Factorials:** The calculus under the hood introduces $\pi$ when evaluating half-steps. For example: 
  $$\left(\frac{1}{2}\right)! = \Gamma\left(\frac{3}{2}\right) = \frac{\sqrt{\pi}}{2} \approx 0.8862$$
* **Negative Fractions:** While negative *integers* break because they hit vertical asymptotes (poles going to infinity), negative *fractions* (like $-0.5$) fall safely between these poles and can be computed using the reflection formula: 
  $$(-0.5)! = \sqrt{\pi} \approx 1.772$$

```python
# =================================================
# MODULE :1 FACTORIAL
# =================================================

def factorial(n):
    
    #Edge cases
    if not isinstance(n,int):
        raise TypeError(f"Expected int, got {type(n).__name__}")
    if n<0:
        raise ValueError("Factorial is not defined for negative numbers!")
    
    fact=1
    for i in range(2,n+1):
        fact*=i   
    return fact

# ==================================================
# TEST CASES
# ==================================================

test_cases =[ (5,120), #positive number
             (0,1),    #Edge case
             (-5,ValueError), #should raise value error
             ("6",TypeError), #should raise type error
             (3.5,TypeError) #should raise type error 
    
]

print("Running Test Cases.........")

for i,(inp,expected) in enumerate(test_cases,1):
    
    try:
        result=factorial(inp)
        if result == expected:
            print(f"Test {i}:- Factorial({inp})! ={result}")
        else:
            print(f"Test {i}:- Factorial({inp})! returned {result}, expected {expected}")
    except Exception as e:
        if isinstance(e,expected):
            print(f"Test {i}:- Factorial({inp}) correctly raised {expected.__name__}")
        else:
            print(f"Test {i}:- Factorial({inp}) raised {type(e).__name__} , expected {expected.__name__}")
          

    
```

# Permutations

#### 1. Core Intuition
A permutation represents an arrangement of items where **order is critically important**. Changing the order of the exact same elements generates a completely brand-new, unique outcome (e.g., the passcode `1-2-3` is fundamentally different from `3-2-1`).

#### 2. Mathematical Definition
When selecting a subset of $r$ items out of a total pool of $n$ distinct items, the formula is:
$$P(n, r) = \frac{n!}{(n-r)!}$$

*The denominator $(n-r)!$ is mathematically structured to cancel out the unused choices from the total descending factorial chain.*

#### 3. Edge Cases & Special Rules
* **Arranging All Items ($r = n$):** When you arrange every single available item, the formula naturally collapses down to a basic factorial:
  $$P(n, n) = \frac{n!}{(n-n)!} = \frac{n!}{0!} = \frac{n!}{1} = n!$$
* **Selecting Zero Items ($r = 0$):** There is exactly $1$ way to arrange zero items from a pool: 
  $$P(n, 0) = \frac{n!}{n!} = 1$$
* **Invalid Constraints ($r > n$):** You cannot physically arrange more items than you possess in your pool. If $r > n$, the value is $0$.

# Combinations

#### 1. Core Intuition
A combination represents a selection of items where **order does not matter at all**. Rearranging the chosen items does not change the outcome; they form the exact same group (e.g., a card hand containing the Ace of Spades and King of Hearts is the exact same hand if you hold the King on the left and the Ace on the right).

#### 4. Mathematical Definition
When choosing a group size of $r$ items out of a total pool of $n$ distinct items, the formula is:
$$C(n, r) = \binom{n}{r} = \frac{n!}{r!(n-r)!}$$

#### 3. Edge Cases & Special Rules
* **Complementary Symmetry:** Choosing a group of $r$ items to *keep* automatically isolates a group of $(n-r)$ items to *leave behind*. Because of this, the combinations are perfectly mirrored:
  $$\binom{n}{r} = \binom{n}{n-r}$$
* **Selecting Everything or Nothing ($r = n$ or $r = 0$):** There is only $1$ way to select all items, and only $1$ way to select nothing:
  $$\binom{n}{n} = 1 \quad \text{and} \quad \binom{n}{0} = 1$$


## The Master Relationship

#### 1. Core Intuition
Permutations and combinations are tethered together by the exact number of ways the selected items can be rearranged ($r!$). A permutation is simply a combination whose selected items have been shaken up and re-ordered.

#### 2. The Algebraic Bridges
To convert a combination into a permutation, multiply the unique groups by the total ways to arrange the chosen subset:
$$P(n, r) = C(n, r) \times r!$$

Conversely, to convert a permutation back into a combination, divide out the redundant, duplicated arrangements caused by order shuffling:
$$C(n, r) = \frac{P(n, r)}{r!}$$



```python
# =================================================
# MODULE :3 COMBINATION
# =================================================


def combination(n,r):
    if (not isinstance(n,int)) or (not isinstance(r,int)):
        raise TypeError("Expected Integer value")
    if n<0 or r<0:
        raise ValueError("Expected positive value(s)")
    if r>n:
        raise ValueError("Combination is Undefined when r>n")
    
    comb = factorial(n)//(factorial(r)*factorial(n-r))
    return comb

# ==================================================
# TEST CASES
# ==================================================


#(Total items , slots to fill)
Ctest_cases=[ (10,3), 
            (5,5),
            (5,6), #should raise error
            (10,-5), #Can't find perm
            (5,0)
               
]

print("TEST CASES:-\n","-"*40)
for i,(n,r) in enumerate(Ctest_cases,1):
    try:
        Cresult=combination(n,r)
        print(f"Test {i} : C{n,r}={Cresult}")
        #Check reult
        print("Is correct? :",np.isclose(Cresult,math.comb(n,r)))
    except Exception as e:
        print(f"Test {i} :",e)
    print("-"*40)        
        


```
